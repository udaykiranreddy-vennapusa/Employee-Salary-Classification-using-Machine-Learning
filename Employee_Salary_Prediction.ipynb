```python
data = data[(data['age']<=75) & (data['age']>=17)]
data.shape
```

```python
data.drop(columns=['education'], inplace=True)
data.columns
```

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
data['workclass']=encoder.fit_transform(data['workclass'])
data['marital-status']=encoder.fit_transform(data['marital-status'])
data['occupation']=encoder.fit_transform(data['occupation'])
data['relationship']=encoder.fit_transform(data['relationship'])
data['race']=encoder.fit_transform(data['race'])
data['gender']=encoder.fit_transform(data['gender'])
data['native-country']=encoder.fit_transform(data['native-country'])
data
```

```python
#spliting into X(independent variable) & Y(dependent variable)
```

```python
X = data.drop(columns=['income'])
Y = data['income']
X
```

```python
Y
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
X = scaler.fit_transform(X)
X
```

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest = train_test_split(X,Y,test_size=0.2,random_state=23,stratify=Y)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier()
knn.fit(xtrain,ytrain)
predict=knn.predict(xtest)
predict
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(ytest,predict)
```

```python
from sklearn.neural_network import MLPClassifier
clf =  MLPClassifier(solver='lbfgs', alpha=1e-5,hidden_layer_sizes=(5,2), max_iter=2000, random_state=1)
clf.fit(xtrain,ytrain)
predict1=clf.predict(xtest)
predict
```

```python
accuracy_score(ytest,predict1)
```

```python
from sklearn.linear_model import LogisticRegression
rgs = LogisticRegression()
rgs.fit(xtrain,ytrain)
predict2=rgs.predict(xtest)
predict2
```

```python
accuracy_score(ytest,predict2)
```
